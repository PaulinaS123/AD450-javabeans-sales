## JavaBeans Sales Analysis
### Part 1 – Construct the DataFrame

This notebook analyzes JavaBeans beverage sales for 2021 and 2022 using Pandas.

```python
import pandas as pd
```

```python
# Create the sales data
sales_data = {
    "Americano": [5628, 6239],
    "Latte": [12426, 25384],
    "Espresso": [852, 756],
    "Cappuccino": [8522, 7561]
}

# Create the DataFrame with the required row labels
coffee_sales = pd.DataFrame(
    sales_data,
    index=["2021 Sales", "2022 Sales"]
)

# Display the DataFrame
coffee_sales
```

### Part 2 – Data Manipulation

```python
# Calculate total sales for each beverage across both years
total_sales = coffee_sales.sum(axis=0)

# Display the results
print("Total sales by beverage:")
display(total_sales)
```

```python
# Calculate the percentage growth from 2021 to 2022
yoy_growth = (
    (coffee_sales.loc["2022 Sales"] - coffee_sales.loc["2021 Sales"])
    / coffee_sales.loc["2021 Sales"]
) * 100

# Display the results rounded to two decimal places
print("Year-over-year growth (%):")
display(yoy_growth.round(2))
```

### Part 3 – Business Insights & Strategy

Best-Selling and Least-Selling Beverages

```python
# Find the best-selling and least-selling beverage for each year
for year in coffee_sales.index:
    best_seller = coffee_sales.loc[year].idxmax()
    least_seller = coffee_sales.loc[year].idxmin()

    print(year)
    print("Best-selling beverage:", best_seller)
    print("Least-selling beverage:", least_seller)
    print()
```

### Strategic Analysis

Based on discovery of sales for Javabeans coffee company, the best selling beverage was the latte growing from 2021 to 2022 grew 104.80%. The company should spend more money on advertising Latte to attract more customers. Americano sales also went up, so JavaBeans should continue selling and promoting it.

Espresso and Cappuccino sales went down in 2022. Espresso had the lowest sales in both years, so the company should look at ways to improve it, such as changing the price or offering promotions. JavaBeans coffee company should not remove these drinks right away. Instead, it should try to improve sales with different marketing strategies and see if customers respond before deciding to remove them from the menu.
